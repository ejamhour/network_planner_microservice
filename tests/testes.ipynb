```python
import requests, base64
from io import BytesIO
from PIL import Image

resp = requests.get("http://127.0.0.1:8080/get_plot?width=800&height=600")

if resp.status_code == 200:
    data = resp.json()
    img_data = base64.b64decode(data["plot"])
    Image.open(BytesIO(img_data)).show()
else:
    print("Error:", resp.status_code)
    print("Detail:", resp.json().get("detail"))
```
