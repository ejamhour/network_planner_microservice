```python
import time
import os
import psutil
import rasterio
from rasterio.windows import Window

def get_process_memory():
    process = psutil.Process(os.getpid())
    return process.memory_info().rss / 1024 / 1024  # Returns MB

def test_keep_open(tile_paths, iterations):
    """Strategy 1: Keep sources open in a list/dictionary."""
    print(f"\n--- Testing 'Keep Open' Strategy ({iterations} iterations) ---")
    mem_init = get_process_memory()
    start_time = time.perf_counter()
    
    # Open once
    srcs = [rasterio.open(p) for p in tile_paths]
    mid_row, mid_col = 12500, 12500
    
    for _ in range(iterations):
        for src in srcs:
            # Read a 1x1 window at the middle of the tile
            _ = src.read(1, window=Window(mid_col, mid_row, 1, 1))
    
    duration = (time.perf_counter() - start_time) * 1000
    mem_final = get_process_memory()
    
    for s in srcs:
        s.close()
        
    print(f"Time: {duration:.2f} ms")
    print(f"Memory Growth: {mem_final - mem_init:.2f} MB")
    return duration

def test_replace_source(tile_paths, iterations):
    """Strategy 2: Open and close the source for every single read."""
    print(f"\n--- Testing 'Replace Source' Strategy ({iterations} iterations) ---")
    mem_init = get_process_memory()
    start_time = time.perf_counter()
    
    mid_row, mid_col = 12500, 12500
    
    for _ in range(iterations):
        for p in tile_paths:
            # Open, read, and close immediately
            with rasterio.open(p) as src:
                _ = src.read(1, window=Window(mid_col, mid_row, 1, 1))
                
    duration = (time.perf_counter() - start_time) * 1000
    mem_final = get_process_memory()
    
    print(f"Time: {duration:.2f} ms")
    print(f"Memory Growth: {mem_final - mem_init:.2f} MB")
    return duration

# Example usage with two of your tiles:
paths = ["curitiba_test/tile_Nyk7pUTx3YY.tif", "curitiba_test/tile_MdCpd97S-IM.tif"]
test_keep_open(paths, 100)
test_replace_source(paths, 100)
```

```python
import ee
import pandas as pd

ee.Initialize(project='ee-ejamhour')

# 1. Define a start and end point for a profile (e.g., West to East)
lon_start, lon_end = -49.32, -49.20
lat_fixed = -25.43
num_points = 10 # Let's start with 10 points

# 2. Create the array of points
points_list = []
for i in range(num_points):
    lon = lon_start + (lon_end - lon_start) * (i / (num_points - 1))
    points_list.append(ee.Geometry.Point([lon, lat_fixed]))

# 3. Get the Dynamic World Image (2024)
dw_image = ee.ImageCollection('GOOGLE/DYNAMICWORLD/V1') \
    .filterBounds(ee.Geometry.MultiPoint(points_list)) \
    .filterDate('2024-01-01', '2024-12-31') \
    .first() \
    .select('label')

# 4. Extract values for all points in the array
def get_val(p):
    return dw_image.sample(p, 10).first().get('label')

# This pulls the data from the cloud for the whole array
results = [get_val(p).getInfo() for p in points_list]

# 5. Display the findings
for i, val in enumerate(results):
    print(f"Point {i}: Lon {points_list[i].coordinates().get(0).getInfo():.4f} -> Class {val}")
```

```python
import requests
from pystac_client import Client
import planetary_computer as pc

# 1. Connect to the Microsoft Catalog
catalog = Client.open(
    "https://planetarycomputer.microsoft.com/api/stac/v1",
    headers={"User-Agent": "CuritibaProject-Research"}
)

# 2. Define the exact area for Curitiba
# Using the same coordinates you provided earlier
aoi = {
    "type": "Polygon",
    "coordinates": [ new ]
}

# 3. Search for the 2023 Esri Land Cover release
# Note: This is a static collection, so the date is an interval for the year
search = catalog.search(
    filter_lang="cql2-json",
    filter={
        "op": "and",
        "args": [
            {"op": "s_intersects", "args": [{"property": "geometry"}, aoi]},
            {"op": "anyinteracts", "args": [{"property": "datetime"}, {"interval": ["2023-01-01T00:00:00Z", "2023-12-31T23:59:59Z"]}]},
            {"op": "=", "args": [{"property": "collection"}, "io-lulc-9-class"]}
        ]
    }
)

# 4. Grab the first result (The tile covering Curitiba)
items = list(search.get_items())
if not items:
    print("No items found. Try a wider date range.")
else:
    first_item = items[0]
    
    # 5. Sign the URL (Microsoft requires a temporary token for access)
    signed_item = pc.sign_item(first_item)
    asset_url = signed_item.assets["data"].href

    # 6. Download the real file locally
    print(f"File found! Downloading from: {asset_url[:50]}...")
    
    filename = "curitiba_esri_static_2023.tif"
    response = requests.get(asset_url, stream=True)
    
    if response.status_code == 200:
        with open(filename, "wb") as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
        print(f"SUCCESS! File saved as {filename}")
    else:
        print("Download failed. Check your internet connection.")
```
