# RPL Points Example With Real Geo Metrics

This notebook uses `points.csv` and runs a simple RPL association test.

Assumptions:

- `itallia` and `torre` are fixed tower roots with rank `0`.
- All other sites are field nodes.
- All nodes use one omni interface.
- Field nodes cannot route, so `relay=False`.
- Candidate links exist only from towers to field nodes.
- Edge metrics are computed by `GraphPlanner` using `AsyncGeoServicePool` and a metric profile selected by interface `tech`.

This notebook requires the planning/geo service to be reachable from the Jupyter environment.


```python
import cisei_lib.planners.geo_rpl_agnostic as grpl
import cisei_lib.planners.planner_classes as pc
import cisei_lib.planners.graph_planner as gp
import importlib
importlib.reload(grpl)
importlib.reload(pc)
importlib.reload(gp)

%matplotlib inline

from math import inf
from pathlib import Path

import pandas as pd

from cisei_lib.io.async_geo_service_pool import AsyncGeoServicePool
from cisei_lib.planners.graph_planner import GraphPlanner
from cisei_lib.planners.planner_classes import AntennaSpec

WORKING_CRS = "EPSG:31982"
TOWER_IDS = {"itallia", "torre"}
RELAY_IDS = {"deneka", "saturno"}
GEO_BASE_URL = "http://planning-service:8080"
GEO_USER_PREFIX = "rpl-real"
GEO_POOL_SIZE = 2
DEFAULT_TECH = "rpl-demo"
DEFAULT_FREQ_MHZ = 900.0
DEFAULT_TX_POWER_DBM = 20.0
DEFAULT_TOWER_TX_POWER_DBM = 43.0
DEFAULT_ANTENNA_ID = "omni_demo"
RELAY_DEGREE = 3
METRIC_SPECS_BY_TECH = {
    DEFAULT_TECH: "default",
}

csv_path = Path("points.csv")
df = pd.read_csv(csv_path)
df["id"] = df["id"].str.strip()
df
```

## 1. Create Site Nodes

Each CSV row becomes one site with one omni radio interface.

Tower roots:

- `fixed=True`
- `rank=0.0`
- `relay=True`

Field nodes:

- `fixed=False`
- `rank=inf`
- `relay=False`

```python
antenna_catalog = {
    DEFAULT_ANTENNA_ID: AntennaSpec(
        kind="omni",
        gain_dbi=0.0,
    )
}

records = df.copy()
records["fixed"] = records["id"].isin(TOWER_IDS)
records["relay"] = records["fixed"] | records["id"].isin(RELAY_IDS)
records["rank"] = records["fixed"].apply(lambda fixed: 0.0 if fixed else inf)
records["tx_power_dbm"] = records["fixed"].apply(
    lambda fixed: DEFAULT_TOWER_TX_POWER_DBM if fixed else DEFAULT_TX_POWER_DBM
)

planner_data = GraphPlanner(
    working_crs=WORKING_CRS,
    antenna_catalog=antenna_catalog,
)

planning_nodes = planner_data.add_records(
    records,
    defaults={
        "tech": DEFAULT_TECH,
        "freq_mhz": DEFAULT_FREQ_MHZ,
        "antenna_id": DEFAULT_ANTENNA_ID,
    },
)
```

## 2. Inspect Normalized Configuration

This table is the planner-facing configuration after defaults, role flags, devices, interfaces, and antenna references are resolved.


```python
config_df = planner_data.node_config_table()
config_df
```

## 3. Convert Interfaces to RPL Nodes

RPL sees interfaces, not physical sites. Each interface becomes one `RPLNode`.

```python
rpl_table = planner_data.rpl_node_table()
rpl_table
```

## 4. Create Candidate Edges

This test offers:

- all compatible tower-to-field links
- nearest-neighbor field links around relay-capable nodes

Edges where both endpoints have `relay=False` are rejected.


```python
planner_data.candidate_edges_roots_plus_nearest_relays(
    degree=RELAY_DEGREE,
    max_distance_m=None,
)

print("candidate edges:", len(planner_data.candidate_edges))

planner_data.candidate_edge_table()
```

## 5. Plot Candidate Graph

This is the candidate graph offered to RPL before edge metrics are computed.


```python
planner_data.show_candidate_graph(
    label_mode="index",
)
```

## 6. Compute Real Edge Metrics

`GraphPlanner.compute_edge_metrics(...)` evaluates every candidate link.

For each link it:

1. Validates compatible `tech` and `freq_mhz`.
2. Gets sites and mount heights from the generated RPL metadata.
3. Calls `AsyncGeoServicePool.get_features(...)`.
4. Selects the compiled metric function by interface `tech`.
5. Stores edge metrics and feature dictionaries on `planner_data`.


```python
geo = AsyncGeoServicePool(
    base_url=GEO_BASE_URL,
    user_prefix=GEO_USER_PREFIX,
    pool_size=GEO_POOL_SIZE,
    timeout=120,
)

edge_metrics = await planner_data.compute_edge_metrics(
    geo,
    metric_specs_by_tech=METRIC_SPECS_BY_TECH,
    manage_geo=True,
)

print("computed edge metrics:", len(edge_metrics))
```

```python
metric_table = planner_data.edge_metric_table()
metric_table
```

## 7. Run RPL

```python
planner_data.run_rpl()
planner_data.rpl_result_counts()
```
