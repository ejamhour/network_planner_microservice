```python
import rssi_core as rc
```

```python
# 1. REMOVE INVALID RECORDS

records = rc.load_json('all_chunks.json')

records = [
    r for r in records
    if isinstance(r.get("features"), dict)
]

rules = [
    {"feature_path": "measures.rssi", "min_value": -120, "max_value": -35, "include_max": True},
    {"feature_path": "tx.pw", "min_value": 0, "max_value": 40, "include_max": True},
    {"feature_path": "tx.ant_gain", "min_value": 0, "max_value": 40, "include_max": True},
    {"feature_path": "rx.ant_gain", "min_value": 0, "max_value": 40, "include_max": True},
    # {"feature_path": "features.tx_near_terminal_clearance_m", "min_value": -20, "max_value": 20, "include_max": True}
]

clean = rc.filter_records_by_many(records, rules)
print("clean:", len(clean))
```

```python
print(clean[0])
```

```python
# 2. ADJUST NO CLUTTER MODEL

rules = [
    {"feature_path": "features.vegetation.boundary", "min_value": 0, "max_value": 1, "include_max": False},
    {"feature_path": "features.buildings.boundary", "min_value": 0, "max_value": 1, "include_max": False},
    {"feature_path": "features.terrain.boundary", "min_value": 0, "max_value": 1, "include_max": False}
]

no_clutter = rc.filter_records_by_many(clean, rules)
print("build_tail:", len(no_clutter))

# No correction model used for comparison
base_model_uncorrected = rc.fit_expression_model(
    no_clutter,
    fixed_terms={
        "tx_power": "tx.pw",
        "tx_gain": "tx.ant_gain",
        "rx_gain": "rx.ant_gain",
        "fspl": "-features.fspl",
        "diff": "-features.delta_diffra",
    },
    fixed_coeffs={
        "tx_power": 1.0,
        "tx_gain": 1.0,
        "rx_gain": 1.0,
        "fspl": 1.0,
        "diff": 1.0,
    },
    learned_terms={},
    y_expr="measures.rssi",
    fit_intercept=True,
    tau=0.5,
)


base_model = rc.fit_expression_model(
    no_clutter,
    fixed_terms={
        "tx_power": "tx.pw",
        "tx_gain": "tx.ant_gain",
        "rx_gain": "rx.ant_gain"
    },
    fixed_coeffs={
        "tx_power": 1.0,
        "tx_gain": 1.0,
        "rx_gain": 1.0,
    },
    learned_terms={
        "fspl": "-features.fspl",
        "diff": "-features.delta_diffra"
    },
    learned_priors={
        "fspl": 1.0,
        "diff": 1.0,
    },
    bounds={
        "fspl": (0.0, None),
        "diff": (0.0, None)       
    },
    y_expr="measures.rssi",
    fit_intercept=False,
    tau=0.6, # increase to make the model optimistic
    l2_to_prior=0.01,
)

#print("intercept:", baseline_model_corrected.intercept)
print("coef corrected:", base_model.learned_coeffs)

#print("training:", baseline_model_corrected.training_info)

print('without correction', rc.evaluate_model(base_model_uncorrected, no_clutter))
print('with correction', rc.evaluate_model(base_model, no_clutter))



# PREDICTION DEMONSTRATION
link = no_clutter[0]

#print(base_model.explain(no_clutter[0]))

print('residual = measured - predicted => negative bias is otimistic')
print(f"link predicition: {base_model_uncorrected.predict(link)} -> {base_model.predict(link)}" )
print(f"link residual: {base_model_uncorrected.residual(link)} -> {base_model.residual(link)}" )

res1 = [ base_model_uncorrected.residual(link) for link in no_clutter ]
res2 = [ base_model.residual(link) for link in no_clutter ]

print('uncorrected model optmistic predictions', sum(x < -10 for x in res1))
print('base model optimistic predictions', sum(x < -10 for x in res2))
```

```python
# 3. EVALUATE BASE MODEL FOR CLUTTER AND NON-CLUTTER

print("NO CLUTTER")
res = rc.describe_residuals(base_model_uncorrected, no_clutter)
res1 = rc.describe_residuals(base_model, no_clutter)

print('Top (10%) optimistic and pessimistic errors')
print('uncorreted:', 'Optimism:', res['10%'], 'Pessimism', res['90%'])
print('base model:', 'Optimism:', res1['10%'], 'Pessimism', res1['90%'])

clutter = [ l for l in clean if l not in no_clutter ]
print("CLUTTER LINKS")
res = rc.describe_residuals(base_model_uncorrected, clutter)
res1 = rc.describe_residuals(base_model, clutter)

print('Top (10%) optimistic and pessimistic errors')
print('uncorreted:', 'Optimism:', res['10%'], 'Pessimism', res['90%'])
print('base model:', 'Optimism:', res1['10%'], 'Pessimism', res1['90%'])
```

```python
# 4. CLUTTER MODEL FOR VEGETATION


rules = [
    {"feature_path": "features.vegetation.boundary", "min_value": 1, "max_value": 10000, "include_max": True},
    {"feature_path": "features.buildings.boundary", "min_value": 0, "max_value": 1, "include_max": False},
    {"feature_path": "features.terrain.fresnel", "min_value": 0, "max_value": 1, "include_max": False}
]

veg_clutter = rc.filter_records_by_many(clean, rules)
print("veg_clutter:", len(veg_clutter))


base_weights = base_model.learned_coeffs

veg_model = rc.fit_expression_model(
    veg_clutter,
    fixed_terms={
        "tx_power": "tx.pw",
        "tx_gain": "tx.ant_gain",
        "rx_gain": "rx.ant_gain",
        "fspl": "-features.fspl",
        "diff": "-features.delta_diffra",
    },
    fixed_coeffs={
        "tx_power": 1.0,
        "tx_gain": 1.0,
        "rx_gain": 1.0,
        "fspl": base_weights['fspl'],
        "diff": base_weights['diff'],
    },
    learned_terms={    
    "veg_boundary": "-math.log10(1.0 + min(features.vegetation.boundary, 1000))",
    "veg_fresnel":  "-math.log10(1.0 + min(features.vegetation.fresnel, 1000))",
    "veg_core":     "-math.log10(1.0 + min(features.vegetation.core, 1000))",
    },
    learned_priors={   
        "veg_boundary": 1,     
        "veg_fresnel": 1,
        "veg_core": 1,
    },
    bounds={
        "veg_boundary": (0.0, None),
        "veg_fresnel": (0.0, None),
        "veg_core": (0.0, None),
    },
    y_expr="measures.rssi", # _model.env_excess_loss
    fit_intercept=False,
    tau=0.5,
    l2_to_prior=0.001,
)

print("veg weights:", veg_model.learned_coeffs)

print('Base model', rc.evaluate_model(base_model, veg_clutter))
print('Veg model', rc.evaluate_model(veg_model, veg_clutter))

res = rc.describe_residuals(base_model, veg_clutter)
res1 = rc.describe_residuals(veg_model, veg_clutter)

print('Top (10%) optimistic and pessimistic errors')
print('base model:', 'Optimism:', res['10%'], 'Pessimism', res['90%'])
print('veg model:', 'Optimism:', res1['10%'], 'Pessimism', res1['90%'])
```

```python
# 5. CLUTTER MODEL FOR BUILDING


rules = [
    {"feature_path": "features.vegetation.fresnel", "min_value": 0, "max_value": 1, "include_max": True},
    {"feature_path": "features.buildings.boundary", "min_value": 1, "max_value": 10000, "include_max": False},
    {"feature_path": "features.terrain.fresnel", "min_value": 0, "max_value": 1, "include_max": False}
]

base_weights = base_model.learned_coeffs

bldg_clutter = rc.filter_records_by_many(clean, rules)
print("bldg_clutter:", len(bldg_clutter))

bldg_model = rc.fit_expression_model(
    bldg_clutter,
    fixed_terms={
        "tx_power": "tx.pw",
        "tx_gain": "tx.ant_gain",
        "rx_gain": "rx.ant_gain",
        "fspl": "-features.fspl",
        "diff": "-features.delta_diffra",
    },
    fixed_coeffs={
        "tx_power": 1.0,
        "tx_gain": 1.0,
        "rx_gain": 1.0,
        "fspl": base_weights['fspl'],
        "diff": base_weights['diff'],
    },
    learned_terms={
    "bldg_boundary": "-math.log10(1.0 + features.buildings.boundary)",
    "bldg_fresnel": "-math.log10(1.0 + features.buildings.fresnel)",
    "bldg_core": "-math.log10(1.0 + features.buildings.core)",
    },
    learned_priors={
        "bldg_boundary": 0,
        "bldg_fresnel": 0,
        "bldg_core": 0,
    },
    bounds={
        "bldg_boundary": (0, None),
        "bldg_fresnel": (0, None),
        "bldg_core": (0, None),
    },
    y_expr="measures.rssi", # _model.env_excess_loss
    fit_intercept=False,
    tau=0.4,
    l2_to_prior=0.001,
)

print("bldg weights:", bldg_model.learned_coeffs)

print('Base model', rc.evaluate_model(base_model, bldg_clutter))
print('Bldg model', rc.evaluate_model(bldg_model, bldg_clutter))

res = rc.describe_residuals(base_model, bldg_clutter)
res1 = rc.describe_residuals(bldg_model, bldg_clutter)

print('Top (10%) optimistic and pessimistic errors')
print('base model:', 'Optimism:', res['10%'], 'Pessimism', res['90%'])
print('bldg model:', 'Optimism:', res1['10%'], 'Pessimism', res1['90%'])
```

```python
# 7. FINAL MODEL
base_weights = base_model.learned_coeffs
veg_weights = veg_model.learned_coeffs
bldg_weights = bldg_model.learned_coeffs


fixed_terms = {
    "tx_power": "tx.pw",
    "tx_gain": "tx.ant_gain",
    "rx_gain": "rx.ant_gain",

    "fspl": "-features.fspl",
    "diff": "-features.delta_diffra",

    "veg_boundary": "-math.log1p(min(features.vegetation.boundary,1000))",
    "veg_fresnel": "-math.log1p(min(features.vegetation.fresnel,1000))",
    "veg_core": "-math.log1p(min(features.vegetation.core,1000))",

    "bldg_boundary": "-math.log10(1.0 + features.buildings.boundary)",
    "bldg_fresnel": "-math.log10(1.0 + features.buildings.fresnel)",
    "bldg_core": "-math.log10(1.0 + features.buildings.core)"
}

fixed_coeffs = {
    "tx_power": 1.0,
    "tx_gain": 1.0,
    "rx_gain": 1.0,

    "fspl": base_weights["fspl"],
    "diff": base_weights["diff"],

    "veg_boundary": veg_weights["veg_boundary"],
    "veg_fresnel": veg_weights["veg_fresnel"],
    "veg_core": veg_weights["veg_core"],

    "bldg_boundary": bldg_weights["bldg_boundary"],
    "bldg_fresnel": bldg_weights["bldg_fresnel"],
    "bldg_core": bldg_weights["bldg_core"],
}

combined_model = rc.ExpressionModel(
    fixed_terms=fixed_terms,
    fixed_coeffs=fixed_coeffs,
    learned_terms={},
    learned_coeffs={},
    intercept=0.0,
    y_expr="measures.rssi",
)

r = clean[0]
print(combined_model.predict(r), combined_model.residual(r))
# print(combined_model.explain(r))

print(rc.evaluate_model(combined_model, veg_clutter))
print(rc.evaluate_model(base_model, veg_clutter))

print(rc.evaluate_model(combined_model, bldg_clutter))
print(rc.evaluate_model(base_model, bldg_clutter))

print(rc.evaluate_model(combined_model, no_clutter))
print(rc.evaluate_model(base_model, no_clutter))

print(rc.evaluate_model(combined_model, clean))
print(rc.evaluate_model(base_model, clean))

print(set(fixed_terms) - set(fixed_coeffs))
print(set(fixed_coeffs) - set(fixed_terms))
```

```python
training = [
    r for r in clean
    if rc.eval_expr(r, "measures.rssi") is not None
    and rc.eval_expr(r, "measures.rssi") < -60    
]

final_model = rc.fit_expression_model(
    clean,
    fixed_terms={
        "tx_power": "tx.pw",
        "tx_gain": "tx.ant_gain",
        "rx_gain": "rx.ant_gain",
        "fspl": "-features.fspl",
        "diff": "-features.delta_diffra",
    },
    fixed_coeffs={
        "tx_power": 1.0,
        "tx_gain": 1.0,
        "rx_gain": 1.0,
        "fspl": base_weights["fspl"],
        "diff": base_weights["diff"],
    },
    learned_terms={
        "veg_boundary": "-math.log10(1.0 + features.vegetation.boundary)",
        "veg_fresnel":  "-math.log10(1.0 + features.vegetation.fresnel)",
        "veg_core":     "-math.log10(1.0 + features.vegetation.core)",

        "bldg_boundary": "-math.log10(1.0 + features.buildings.boundary)",
        "bldg_fresnel":  "-math.log10(1.0 + features.buildings.fresnel)",
        "bldg_core":     "-math.log10(1.0 + features.buildings.core)",

    },
    learned_priors={
        "veg_boundary": veg_weights["veg_boundary"],
        "veg_fresnel": veg_weights["veg_fresnel"],
        "veg_core": veg_weights["veg_core"],

        "bldg_boundary": bldg_weights["bldg_boundary"],
        "bldg_fresnel": bldg_weights["bldg_fresnel"],
        "bldg_core": bldg_weights["bldg_core"],

    },
    bounds={
        "veg_boundary": (0.0, None),
        "veg_fresnel": (0.0, None),
        "veg_core": (0.0, None),

        "bldg_boundary": (0.0, None),
        "bldg_fresnel": (0.0, None),
        "bldg_core": (0.0, None),

    },
    fit_intercept=False,
    tau=0.5,
    l2_to_prior=1e-4,
)

print(final_model.learned_coeffs)
```

```python
border_line = [
    r for r in clean
    if rc.eval_expr(r, "measures.rssi") is not None
    and rc.eval_expr(r, "measures.rssi") < -70
    and rc.eval_expr(r, "measures.rssi") > -80
]

bad = [
    r for r in clean
    if rc.eval_expr(r, "measures.rssi") is not None
    and rc.eval_expr(r, "measures.rssi") < -80
    and rc.eval_expr(r, "measures.rssi") > -100
]


for dataset in [clean, border_line, bad] :

    print(rc.evaluate_model(base_model, dataset))
    print(rc.evaluate_model(final_model, dataset))
    print(rc.evaluate_model(base_model_uncorrected, dataset))

    res = rc.describe_residuals(base_model, dataset)
    res1 = rc.describe_residuals(final_model, dataset)
    res2 = rc.describe_residuals(base_model_uncorrected, dataset)


    print('Top (25%) optimistic and pessimistic errors')
    print('base model:', 'Optimism:', res['25%'], 'Pessimism', res['75%'])
    print('final model:', 'Optimism:', res1['25%'], 'Pessimism', res1['75%'])
    print('analytical model:', 'Optimism:', res2['25%'], 'Pessimism', res2['75%'])

```
