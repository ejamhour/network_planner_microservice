# PlanningScenario Tutorial

`PlanningScenario` is the user-facing preparation layer for planning runs. It owns the scenario definition: antenna profiles, interface profiles, device profiles, concrete site instances, connectivity rules, manual candidate edges and metric specs.

It does not compute geo features, metrics, RPL, or the final planned network. Those steps belong to `GraphPlanner`, `CellPlanner`, or another specialized planner.

````python
from pathlib import Path

import pandas as pd

from cisei_lib.planners import PlanningScenario

OUT_DIR = Path("scenario_exports")
OUT_DIR.mkdir(exist_ok=True)
````

## 1. Create an Empty Scenario

`working_crs` is the projected CRS used for UTM-like coordinates and distance-based planning. If records contain only `lat`/`lon`, the scenario can resolve `x`/`y` using this CRS when site instances are created.

````python
scenario = PlanningScenario(working_crs="EPSG:31982")
scenario.working_crs
````

## 2. Add Antenna Profiles

Antenna profiles are referenced by interfaces through `antenna_id`.

Required fields are intentionally minimal. Common fields are:

- `kind`: usually `omni`, `sector`, or `directional`
- `model`: optional model/name
- `gain_dbi`: antenna gain
- `azimuth_deg`: sector/directional pointing angle, degrees clockwise from north
- `beamwidth_deg`: angular opening for sector/directional candidate filtering
- `max_links`: optional capacity annotation

````python
scenario.add_antenna_profile("lte_client_omni", {
    "kind": "omni",
    "model": "client_omni",
    "gain_dbi": 6.0,
})

scenario.add_antenna_profile("lte_sector_0", {
    "kind": "sector",
    "model": "sector_120_0",
    "gain_dbi": 14.0,
    "azimuth_deg": 0.0,
    "beamwidth_deg": 120.0,
})

scenario.add_antenna_profile("lte_sector_120", {
    "kind": "sector",
    "model": "sector_120_120",
    "gain_dbi": 14.0,
    "azimuth_deg": 120.0,
    "beamwidth_deg": 120.0,
})

scenario.add_antenna_profile("lte_sector_240", {
    "kind": "sector",
    "model": "sector_120_240",
    "gain_dbi": 14.0,
    "azimuth_deg": 240.0,
    "beamwidth_deg": 120.0,
})

scenario.profile_table("antennas")
````

## 3. Add Interface Profiles

Interface profiles define the radio interface that will become an RPL node later.

Required fields:

- `tech`
- `freq_mhz`
- `tx_power_dbm`
- `antenna_id`

Optional fields:

- `can_relay`: whether this interface can forward in the RPL graph for this technology
- `medium`: usually `rf`

````python
scenario.add_interface_profile("lte_sector_0", {
    "tech": "lte",
    "freq_mhz": 400.0,
    "tx_power_dbm": 43.0,
    "antenna_id": "lte_sector_0",
    "can_relay": False,
})

scenario.add_interface_profile("lte_sector_120", {
    "tech": "lte",
    "freq_mhz": 400.0,
    "tx_power_dbm": 43.0,
    "antenna_id": "lte_sector_120",
    "can_relay": False,
})

scenario.add_interface_profile("lte_sector_240", {
    "tech": "lte",
    "freq_mhz": 400.0,
    "tx_power_dbm": 43.0,
    "antenna_id": "lte_sector_240",
    "can_relay": False,
})

scenario.add_interface_profile("lte_client", {
    "tech": "lte",
    "freq_mhz": 400.0,
    "tx_power_dbm": 20.0,
    "antenna_id": "lte_client_omni",
    "can_relay": False,
})

scenario.profile_table("interfaces")
````

## 4. Add Device Profiles

Device profiles define one installed device and the interfaces it contains.

Required field:

- `interfaces`: non-empty list of interface profile ids

Common optional fields:

- `connected`: already connected to the backbone or an accepted existing network
- `can_route`: can route between interfaces in the same device
- `rank`: initial RPL rank, usually `0.0` for connected devices
- `mount_height_m`: installation height in meters

````python
scenario.add_device_profile("cell_sector_3x120", {
    "kind" : "tower",
    "connected": True,
    "can_route": True,
    "rank": 0.0,
    "mount_height_m": 100.0,
    "interfaces": ["lte_sector_0", "lte_sector_120", "lte_sector_240"],
})

scenario.add_device_profile("lte_leaf", {
    "connected": False,
    "can_route": False,
    "mount_height_m": 7.0,
    "interfaces": ["lte_client"],
})

scenario.profile_table("devices")
````

## 5. Instantiate Sites From Records

File reading stays outside the library. A notebook may read CSV, Excel, GeoJSON, or create a DataFrame manually. The scenario consumes records.

The standard instance table uses one row per site. Each row provides an id (`site_id`, `position_id`, `id`, or `name`), either `lat`/`lon` or `x`/`y`, and a `device_profile` column that selects which registered device profile will be instantiated.

Row values may override profile defaults such as `connected`, `can_route`, `rank`, `kind`, and `mount_height_m`.


````python
points = pd.DataFrame([
    {"site_id": "torre", "lat": -25.0000, "lon": -49.0000, "device_profile": "cell_sector_3x120"},
    {"site_id": "itallia", "lat": -25.0040, "lon": -49.0180, "device_profile": "cell_sector_3x120"},
    {"site_id": "woodland", "lat": -25.0100, "lon": -49.0100, "device_profile": "lte_leaf"},
    {"site_id": "saturno", "lat": -25.0140, "lon": -49.0200, "device_profile": "lte_leaf"},
    {"site_id": "deneka", "lat": -25.0170, "lon": -49.0150, "device_profile": "lte_leaf"},
])

scenario.add_node_instances(points)

scenario.instance_table()
````

## 6. Add Metric Specs And Connectivity Rules

Metric specs are mapped by technology. Connectivity rules are stored in the scenario and interpreted later by a graph builder.

Manual candidate edges can also be added. They are candidates RPL may use; they do not force the final planned result.

````python
scenario.add_metric_spec("lte", {"spec": "lte_tower_node"})

scenario.add_connectivity_rule({
    "kind": "rf",
    "tech": "lte",
    "source": "connected",
    "destination": "unconnected",
})

scenario.add_candidate_edge(
    "torre:d0:i1",
    "woodland:d0:i0",
    attrs={"kind": "rf", "tech": "lte", "note": "manual candidate for testing"},
)

scenario.candidate_edge_table()
````

## 7. Validate And Save A TOML/CSV Bundle

`validate()` returns a list of problems instead of raising. An empty list means the cross-references known by the scenario are consistent.

`save_bundle()` writes the reusable configuration to TOML. If concrete node instances exist, it also writes a companion CSV with the same stem and adds an `[instances]` section to the TOML. This keeps the profile configuration readable and keeps site instances editable as a normal table.


````python
errors = scenario.validate()
errors
````

````python
out_path = OUT_DIR / "planning_scenario_lte_sector_demo.toml"
toml_path, instances_path = scenario.save_bundle(out_path)

print("TOML:", toml_path.resolve())
print("instances CSV:", instances_path.resolve() if instances_path else None)
print(toml_path.read_text(encoding="utf-8")[:1500])
````

## 8. Read The Scenario Back From TOML

Reading TOML calls the same scenario methods. If the TOML has `[instances]` and the companion CSV exists, `from_toml(..., load_instances=True)` imports the instances automatically.

Embedded `[[sites]]` records are still supported for exact object serialization, but the profile + CSV bundle is the cleaner notebook workflow.


````python
loaded = PlanningScenario.from_toml(out_path)

print("validation errors:", loaded.validate())
print("instance source:", loaded.instance_source)
loaded.instance_table().head()
````

## 9. Multi-Interface Scenario: LTE + Node-to-Node

This scenario gives each field device two interfaces:

- `i0`: LTE access interface
- `i1`: node-to-node interface

Relay devices can route between interfaces. Leaf devices have both interfaces available, but do not route between them.

````python
hybrid = PlanningScenario(working_crs="EPSG:31982")

hybrid.add_antenna_profile("lte_omni", {
    "kind": "omni",
    "model": "lte_omni",
    "gain_dbi": 6.0,
})
hybrid.add_antenna_profile("n2n_omni", {
    "kind": "omni",
    "model": "n2n_omni",
    "gain_dbi": 3.0,
})

hybrid.add_interface_profile("lte_access", {
    "tech": "lte",
    "freq_mhz": 400.0,
    "tx_power_dbm": 20.0,
    "antenna_id": "lte_omni",
    "can_relay": False,
})
hybrid.add_interface_profile("n2n_leaf", {
    "tech": "n2n",
    "freq_mhz": 915.0,
    "tx_power_dbm": 20.0,
    "antenna_id": "n2n_omni",
    "can_relay": False,
})
hybrid.add_interface_profile("n2n_relay", {
    "tech": "n2n",
    "freq_mhz": 915.0,
    "tx_power_dbm": 23.0,
    "antenna_id": "n2n_omni",
    "can_relay": True,
})

hybrid.add_device_profile("tower_lte", {
    "connected": True,
    "can_route": True,
    "rank": 0.0,
    "mount_height_m": 100.0,
    "interfaces": ["lte_access"],
})
hybrid.add_device_profile("field_leaf_dual", {
    "connected": False,
    "can_route": False,
    "mount_height_m": 7.0,
    "interfaces": ["lte_access", "n2n_leaf"],
})
hybrid.add_device_profile("field_relay_dual", {
    "connected": False,
    "can_route": True,
    "mount_height_m": 7.0,
    "interfaces": ["lte_access", "n2n_relay"],
})

hybrid_points = pd.DataFrame([
    {"site_id": "torre", "lat": -25.0000, "lon": -49.0000, "device_profile": "tower_lte"},
    {"site_id": "saturno", "lat": -25.0140, "lon": -49.0200, "device_profile": "field_relay_dual"},
    {"site_id": "deneka", "lat": -25.0170, "lon": -49.0150, "device_profile": "field_relay_dual"},
    {"site_id": "woodland", "lat": -25.0100, "lon": -49.0100, "device_profile": "field_leaf_dual"},
])

hybrid.add_node_instances(hybrid_points)

hybrid.node_table()
````

## 10. Connectivity Rules For The Hybrid Scenario

The generic scenario only stores the rules. Later, `GraphPlanner` or a specialization will decide how to generate automatic edges from them.

Here we define:

- LTE access from connected interfaces to unconnected interfaces
- node-to-node RF candidates among relay-capable and client interfaces
- internal links between interfaces in devices that can route

````python
hybrid.add_metric_spec("lte", {"spec": "lte_tower_node"})
hybrid.add_metric_spec("n2n", {"spec": "default"})

hybrid.add_connectivity_rule({
    "kind": "rf",
    "tech": "lte",
    "source": "connected",
    "destination": "unconnected",
})
hybrid.add_connectivity_rule({
    "kind": "rf",
    "tech": "n2n",
    "source": "relay",
    "destination": "any",
    "degree": 3,
    "limit": 1000.0,
})
hybrid.add_connectivity_rule({
    "kind": "internal",
    "enabled_when_device_can_route": True,
    "metric": 0.0,
})

hybrid.add_candidate_edge(
    "saturno:d0:i1",
    "woodland:d0:i1",
    attrs={"kind": "rf", "tech": "n2n", "note": "forced candidate, not forced result"},
)

print("validation errors:", hybrid.validate())
hybrid.candidate_edge_table()
````

## 11. Export The Hybrid Scenario

The hybrid scenario uses the same bundle mechanism. The TOML stores profiles, metrics, connectivity rules, and manual candidate edges. The companion CSV stores the concrete site instances and the profile assigned to each site.


````python
hybrid_path = OUT_DIR / "planning_scenario_hybrid_demo.toml"
hybrid_toml_path, hybrid_instances_path = hybrid.save_bundle(hybrid_path)

print("TOML:", hybrid_toml_path.resolve())
print("instances CSV:", hybrid_instances_path.resolve() if hybrid_instances_path else None)
print(hybrid_toml_path.read_text(encoding="utf-8")[:2200])
````
