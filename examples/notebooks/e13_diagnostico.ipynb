## 7. Aplicar o modelo da E08 e classificar os enlaces

O SDK lê o modelo já treinado e enriched_features.json, compara previsão e medição e grava evaluation.json. Não há novo treinamento nem seleção de caminhos. O limiar de -80 dBm é ilustrativo e deve ser definido pelo engenheiro.

Verde: atendimento satisfatório (previsão e medição boas). Vermelho: investigar instalação ou defeito (previsão boa e medição ruim). Azul: medição satisfatória; sem correção (previsão ruim e medição boa). Marrom: revisar projeto (previsão e medição ruins). As cinco conexões internas não recebem classificação de RSSI. Sem medição, a previsão pode indicar revisão de projeto; uma previsão satisfatória permanece identificada como sem medição.

As antenas YAGI preservam o ganho de 14 dBi, mas seus diagramas e orientações não estão disponíveis. O perfil de ganho constante não afirma que sejam antenas omni. As indicações devem ser lidas considerando essa limitação e os parâmetros documentados do projeto.
