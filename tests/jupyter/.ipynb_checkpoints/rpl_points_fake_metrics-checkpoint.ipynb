# RPL Points Example With Fake Metrics

This notebook uses `points.csv` and runs a simple RPL association test.

Assumptions:

- `itallia` and `torre` are fixed tower roots with rank `0`.
- All other sites are field nodes.
- All nodes use one omni interface.
- Field nodes cannot route, so `relay=False`.
- Candidate links exist only from towers to field nodes.
- Edge metrics are fake for now, based on projected distance.

Later, the fake metric cell can be replaced by a real metric provider using the geo service.

````python

import cisei_lib.planners.geo_rpl_agnostic as grpl
import cisei_lib.planners.planner_classes as pc

import importlib
importlib.reload(grpl)
importlib.reload(pc)

%matplotlib inline

from math import hypot, inf
from pathlib import Path

import pandas as pd

from cisei_lib.planners.geo_rpl_agnostic import GeoRPL
from cisei_lib.planners.planner_classes import (
    AntennaSpec,
    FieldSite,
)

WORKING_CRS = "EPSG:31982"
TOWER_IDS = {"itallia", "torre"}

csv_path = Path("points.csv")
df = pd.read_csv(csv_path)
df["id"] = df["id"].str.strip()
df
````

## 1. Create Site Nodes

Each CSV row becomes one site with one omni radio interface.

Tower roots:

- `fixed=True`
- `rank=0.0`
- `relay=True`

Field nodes:

- `fixed=False`
- `rank=inf`
- `relay=False`

````python
planning_nodes = []

for row in df.itertuples(index=False):
    is_tower = row.id in TOWER_IDS

    node = FieldSite.with_default_interface(
        site_id=row.id,
        lat=float(row.lat),
        lon=float(row.lon),
        tech="rpl-demo",
        fixed=is_tower,
        relay=is_tower is True,
        rank=0.0 if is_tower else inf,
        antenna=AntennaSpec(
            kind="omni",
            gain_dbi=0.0,
            height_m=float(row.ant_h),
        ),
    )
    node.resolve_position(WORKING_CRS)
    planning_nodes.append(node)

for node in planning_nodes:
    interface = node.interfaces[0]
    print(
        f"{node.node_id:10s}",
        f"interface={interface.interface_id:18s}",
        f"fixed={interface.fixed}",
        f"relay={interface.relay}",
        f"rank={interface.rank}",
        f"height={interface.antenna.height_m}",
    )
````

## 2. Convert Interfaces to RPL Nodes

RPL sees interfaces, not physical sites. Each interface becomes one `RPLNode`.

````python
rpl_nodes = []
node_by_site = {}

for node in planning_nodes:
    node_by_site[node.node_id] = node
    rpl_nodes.extend(node.rpl_nodes())

interface_by_site = {
    node.node_id: node.rpl_nodes()[0].node_id
    for node in planning_nodes
}

for rpl_node in rpl_nodes:
    print(
        f"{rpl_node.node_id:18s}",
        f"fixed={rpl_node.fixed}",
        f"relay={rpl_node.relay}",
        f"pos_utm={tuple(round(value, 2) for value in rpl_node.pos_utm)}",
    )
````

## 3. Create Candidate Edges

This first test only offers tower-to-field links to RPL.

Because field nodes are `relay=False`, even if field-to-field edges existed they would not propagate rank. Here we keep the graph intentionally simple.

````python
tower_nodes = [
    node
    for node in planning_nodes
    if node.node_id in TOWER_IDS
]

field_nodes = [
    node
    for node in planning_nodes
    if node.node_id not in TOWER_IDS
]

candidate_edges = []

for tower in tower_nodes:
    tower_interface = interface_by_site[tower.node_id]
    for field in field_nodes:
        field_interface = interface_by_site[field.node_id]
        candidate_edges.append((tower_interface, field_interface))

print("candidate edges:", len(candidate_edges))
candidate_edges[:10]
````

## 4. Compute Fake Edge Metrics

This cell is the placeholder for the real metric provider.

For now:

```text
metric = max(1.0, distance_km)
```

The RPL result will therefore connect each field node to the nearest tower root.

````python
rpl_by_id = {node.node_id: node for node in rpl_nodes}
edge_metrics = {}

for src, dst in candidate_edges:
    sx, sy = rpl_by_id[src].pos_utm
    dx, dy = rpl_by_id[dst].pos_utm
    distance_m = hypot(dx - sx, dy - sy)
    metric = max(1.0, distance_m / 1000.0)
    edge_metrics[(src, dst)] = round(metric, 3)

for edge, metric in list(edge_metrics.items())[:10]:
    print(edge, metric)
````

## 5. Run RPL

````python
planner = GeoRPL()
planner.set_nodes(rpl_nodes)
planner.set_edge_metrics(edge_metrics)
planner.run_RPL()

print("RPL result nodes:", planner.G_res.number_of_nodes())
print("RPL result edges:", planner.G_res.number_of_edges())
````

## 6. Print Results

The user-facing output maps interface-level RPL parents back to physical sites.

````python
index_df = planner.node_index_table(G=planner.G_res)
index_df
````

Summary by physical field node. The `node_index` column is the compact label shown in the graph.

````python
index_df[index_df["fixed"] == False][
    ["node_index", "site_id", "parent_index", "parent_site_id", "rank"]
].sort_values("site_id")
````
