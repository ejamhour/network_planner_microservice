```python
from math import inf
import random

from cisei_lib.planners.geo_rpl_agnostic import GeoRPL
from cisei_lib.planners.planner_classes import RPLNode

random.seed(42)

# One fixed gateway + 10 floating nodes.
nodes = [
    RPLNode(
        node_id="gateway",
        fixed=True,
        relay=True,
        rank=0.0,
        pos_utm=(50.0, 50.0),
    )
]

# Fake grid-like node positions inside 100 x 100.
for i in range(1, 11):
    nodes.append(
        RPLNode(
            node_id=f"node_{i}",
            fixed=False,
            relay=True,
            rank=inf,
            pos_utm=(
                random.randint(0, 100),
                random.randint(0, 100),
            ),
        )
    )

planner = GeoRPL()
planner.set_nodes(nodes)

# Create simple candidate edges:
# - gateway connects to every node
# - each node also connects to the next node
edges = {}

for node in nodes[1:]:
    edges[("gateway", node.node_id)] = random.randint(1, 5)

for i in range(1, 10):
    edges[(f"node_{i}", f"node_{i+1}")] = random.randint(1, 5)

planner.set_edge_metrics(edges)
planner.run_RPL()

print("Nodes:")
for node_id, attrs in planner.G_res.nodes(data=True):
    print(
        f"{node_id:8s} "
        f"pos={attrs['pos']} "
        f"parent={attrs.get('parent')} "
        f"rank={attrs['rank']:.1f}"
    )

print("\nSelected links:")
for src, dst, attrs in planner.G_res.edges(data=True):
    print(f"{src:8s} -- {dst:8s} metric={attrs['metric']}")
```
