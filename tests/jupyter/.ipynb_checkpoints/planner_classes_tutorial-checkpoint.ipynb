# Planner Classes Tutorial

This notebook shows the planning data model used before running RPL.

Core idea:

- A **position** is a physical place with coordinates.
- A **radio** is equipment installed at a position.
- An **interface** is the connectable endpoint of a radio.
- An **antenna** is a property of an interface.
- RPL sees interfaces as graph nodes.

Interface IDs follow:

```text
position:radio:interface
```

Example:

```text
copel:r0:i0
```

````python
from math import inf

from cisei_lib.planners.planner_classes import (
    AntennaSpec,
    FieldNode,
    InterfacePattern,
    NodePattern,
    PlanningInterface,
    PlanningPosition,
    PlanningRadio,
    RadioPattern,
    TowerNode,
    get_utm_epsg,
    make_interface_id,
    make_radio_id,
)
from cisei_lib.planners.geo_rpl_agnostic import GeoRPL

WORKING_CRS = "EPSG:31982"  # SIRGAS 2000 / UTM zone 22S
WORKING_CRS
````

## 1. Create One Wi-SUN Field Node

`FieldNode.with_default_interface(...)` is a convenience constructor. It creates:

- one position
- one radio
- one interface
- one antenna assigned to that interface

````python
copel = FieldNode.with_default_interface(
    position_id="copel",
    lat=-25.432730835150082,
    lon=-49.339207159685344,
    tech="wisun",
    fixed=False,
    relay=True,
    rank=inf,
    antenna=AntennaSpec(
        kind="omni",
        gain_dbi=6.25,
        height_m=7,
    ),
)

copel.resolve_position(WORKING_CRS)

print("position:", copel.position)
print("radio:", copel.radios[0])
print("interface:", copel.interfaces[0])
print("rpl node:", copel.rpl_nodes()[0])
````

## 2. Use Patterns for Repeated Devices

Patterns avoid repeating the same radio/interface/antenna definition for thousands of similar devices.

Pattern keys are local:

```text
r0, i0
```

Instantiated IDs are global:

```text
barigui:r0:i0
```

````python
wisun_pattern = NodePattern.default_wisun(
    antenna=AntennaSpec(
        kind="omni",
        gain_dbi=6.25,
        height_m=7,
    ),
    fixed=False,
    relay=True,
    rank=inf,
)

rows = [
    {"id": "tenis", "lat": -25.425356577135183, "lon": -49.29212222420152},
    {"id": "barigui", "lat": -25.428858965702883, "lon": -49.31129716330312},
    {"lat": -25.427623857466518, "lon": -49.28564299624496},
    {"position_id": "utm_only", "x": 667000.0, "y": 7186000.0},
]

field_nodes = wisun_pattern.instantiate_many(
    rows,
    id_prefix="wisun",
    working_crs=WORKING_CRS,
    resolve=True,
)

for node in field_nodes:
    print(node.node_id, "->", node.rpl_nodes()[0].node_id, node.position.projected_pos)
````

Rows without `position_id`, `id`, or `name` receive automatic IDs using `id_prefix`.

Projected-only rows are allowed, but `x/y` must already be in `WORKING_CRS`.

````python
try:
    wisun_pattern.instantiate_many([
        {"id": "duplicated", "lat": -25.1, "lon": -49.1},
        {"id": "duplicated", "lat": -25.2, "lon": -49.2},
    ])
except ValueError as error:
    print(error)
````

## 3. Build a Tower With Sector Interfaces

A tower may have multiple interfaces at the same position. Each sector is an interface with its own antenna properties.

````python
tower_pattern = NodePattern.tower_sectors(
    sectors=[
        AntennaSpec(kind="sector", gain_dbi=17, height_m=100, azimuth_deg=0, beamwidth_deg=120, max_links=64),
        AntennaSpec(kind="sector", gain_dbi=17, height_m=100, azimuth_deg=120, beamwidth_deg=120, max_links=64),
        AntennaSpec(kind="sector", gain_dbi=17, height_m=100, azimuth_deg=240, beamwidth_deg=120, max_links=64),
    ],
    tech="lte",
    fixed=True,
    relay=True,
    rank=0.0,
)

torre = tower_pattern.instantiate(
    "torre",
    lat=-25.423532434217368,
    lon=-49.29399235240344,
    working_crs=WORKING_CRS,
    resolve=True,
)

for interface in torre.interfaces:
    print(interface.interface_id, interface.antenna.to_dict())
````

The older convenience constructor still works and delegates to the pattern mechanism.

````python
legacy_style_tower = TowerNode.with_sector_interfaces(
    "tower_a",
    lat=-25.40,
    lon=-49.30,
    sectors=[AntennaSpec(kind="sector"), AntennaSpec(kind="sector")],
)

print([interface.interface_id for interface in legacy_style_tower.interfaces])
````

## 4. Manual Construction

Manual construction is useful when a future antenna planner expands logical interfaces into physical radios/interfaces.

````python
position = PlanningPosition(
    position_id="manual_site",
    lat=-25.43,
    lon=-49.30,
)

radio = PlanningRadio(
    radio_id=make_radio_id("manual_site", 0),
    position_id="manual_site",
    tech="backhaul",
)

interface = PlanningInterface(
    interface_id=make_interface_id("manual_site", 0, 0),
    radio_id=radio.radio_id,
    position_id="manual_site",
    antenna=AntennaSpec(kind="directional", gain_dbi=14, height_m=7),
    fixed=False,
    relay=True,
    rank=inf,
)

position.resolve(WORKING_CRS)
interface.to_rpl_node(position, radio)
````

## 5. Convert Interfaces to RPL Nodes

`GeoRPL` receives only simplified `RPLNode` objects and weighted feasible edges. It does not know about sectors, antenna gain, technology, or radio inventory.

````python
gateway_pattern = NodePattern.default_wisun(
    antenna=AntennaSpec(kind="omni", gain_dbi=6.25, height_m=30),
    fixed=True,
    relay=True,
    rank=0.0,
)

gateway = gateway_pattern.instantiate(
    "gateway",
    lat=-25.43017531546412,
    lon=-49.266232140399694,
    working_crs=WORKING_CRS,
    resolve=True,
)

planning_nodes = [gateway, *field_nodes]
rpl_nodes = []

for node in planning_nodes:
    rpl_nodes.extend(node.rpl_nodes())

for node in rpl_nodes:
    print(
        f"{node.node_id:18s}",
        "fixed=", node.fixed,
        "relay=", node.relay,
        "position=", node.extra["position_id"],
        "radio=", node.extra["radio_id"],
        "tech=", node.extra.get("tech"),
    )
````

## 6. Run RPL With Fake Edge Metrics

In production, a graph-preparation class will decide which links are feasible and compute metrics. Here we use simple fake costs to show the data flow.

````python
planner = GeoRPL()
planner.set_nodes(rpl_nodes)

gateway_id = gateway.rpl_nodes()[0].node_id
field_ids = [node.rpl_nodes()[0].node_id for node in field_nodes]

edges = {}

for index, node_id in enumerate(field_ids, start=1):
    edges[(gateway_id, node_id)] = float(index)

for left, right in zip(field_ids, field_ids[1:]):
    edges[(left, right)] = 1.0

planner.set_edge_metrics(edges)
planner.run_RPL()

for node_id, attrs in planner.G_res.nodes(data=True):
    print(
        f"{node_id:18s}",
        "parent=", attrs.get("parent"),
        "rank=", round(attrs["rank"], 3),
    )
````

````python
planner.show_network(G=planner.G_res)
````
