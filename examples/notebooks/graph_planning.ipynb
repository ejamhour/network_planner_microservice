# Wi-SUN mesh planning

Plan a single-radio Wi-SUN mesh with two gateways, routing nodes, and two leaf nodes. Unlike cell planning, routing nodes may forward traffic through multiple hops toward a gateway. `deneka` and `woodland` are leaves and therefore cannot relay traffic.

## 1. Connect to the planning service

```python
import pandas as pd
from cisei_planning_sdk import PlanningClient, PlanningProject, plot_network

client = PlanningClient(
    "http://network-service:8080",
    workspace="..",
    timeout=900,
)
```

## 2. Open the Wi-SUN project

The project contains one 915 MHz radio per device. Gateways and router nodes can relay; leaf nodes cannot.

```python
project = PlanningProject("../projects/graph_planner_wisun")
scenario = client.open_project(project, planner="graph")

scenario.summary
```

## 3. Build candidate mesh links

Each relay-capable node considers nearby Wi-SUN neighbors. Leaves may receive a link but are not used to forward traffic.

```python
candidates = scenario.build_candidate_edges()

pd.DataFrame(candidates["candidate_edges"])
```

## 4. Compute link metrics

The SDK saves reusable geographic features in the project's `features.json`.

```python
metrics = scenario.compute_metrics()

pd.DataFrame(metrics["metrics"])
```

## 5. Solve and evaluate the mesh

```python
solution = scenario.solve()

pd.DataFrame(solution["planned_edges"])
```

```python
evaluation = scenario.evaluate(rank_threshold=10)

evaluation["quality_summary"]
```

## 6. Save and inspect the result

The graphs are generated from the result reloaded from this project's `result.json`.

```python
scenario.save_result(rank_threshold=10)
```

```python
saved_result = scenario.load_result()
plot_network(saved_result, kind="candidate");
```

```python
plot_network(saved_result, kind="final");
```
